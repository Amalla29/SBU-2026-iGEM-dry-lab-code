# ODE simulation lesson (created by Andrew)

An ODE (ordinary differential equations) is an equation that relates an unknown function of a single variable to its derivatives. These equations describe how one variable or quantity changes with respect to another variable, like time or space.

We will be working with multiple ODEs at once, so doing things by hand isn't very practical. Luckily, you guys should be experts at Python already, so we can simulate our ODEs using a bit of coding magic.



# Systems of ODEs

Rather than looking at one ODE at a time, we can create a system of ODEs, similar to those you've probably seen in algebra. These systems will consist of derivatives of several quantities, each of which depend on the same variable (in our case it will be time)

Cosider the following system of ODEs with the following initial conditions:
$$
\frac{dx}{dt}= x-5y \hspace{0.5in} \frac{dy}{dt}= y+5x
$$
$$
( x(0),y(0)) = (5,5)
$$


# Solving our system

# Step 1: Set parameters and initialize variables
In order to solve our system, we need to define a set of parameters and create an array for each variable.

Key parameters we need include: time of simulation, time step

To make arrays for each variable, we will ```numpy``` from last time and use the knowledge we gained from last time.



````python
# Import our needed libraries (MUST BE DONE EVERY TIME)
import numpy as np
import matplotlib.pyplot as plt

# Define parameters: time length and time step (delta_t)

total_time = 5
delta_t = 0.01 # we will see what happens when we change this

# Create arrays for each variable: x, y, and t

t = np.arange(0, total_time, delta_t)
num_step = np.size(t) # we will use this variable later

X = np.zeros(np.size(t))
Y = np.zeros(np.size(t))

# Set up initial conditions
X[0] = 5
Y[0] = 5
````

# Step 2: Integrate

The most effective way to integrate our system in Python is to use the Forward Euler method. You may  have learned this in Calc 2, but here is the procedure:


*   Begin with your starting equation and initial conditions
*   Calculate the derivative of your equation at the initial conditions
* Multiply your derivative by a very short step to estimate the next point
* Use the next point that you estimates as your new starting point
* Repeat

The main formula is as followed:

$$y(t+\Delta t) = y(t) +  \left ( \frac{dy(t)}{dt} \right )\Delta t$$

To do this in our code, we have to use a 'for' loop, which is an iterative loop that will continue to repeat until the conditions are no longer met. The main format of a 'for' loop is as followed:

```for (item) in (set range/sequence):```

The item is the variable that changes with with each loop. We typically use i, but it could be anything. The range is the set of numbers the the loop will go through.

```Example: for i in range(5):```

```print(i)```

This will print numbers 0 through 4 because indexing in Python starts at 0, not 1

```Example 2: DNA = 'ATGCGTACGCCATAT'```

```for base in DNA:```

```print(base)```

This will print all the letters in the DNA sequence we created

```Example 3: X = np.zeros(5)```

```for i in range(4):```

```X[i+1] = X[i] + 3```

This will go through each element in the array and add 3 (Try testing this code out for yourself!)

This last example is what we will use to integrate our ODEs

````python
for i in range(0,num_step-1) :
  dxdt = X[i]-5*Y[i] # derivative of X
  dydt = Y[i]+5*X[i] # derivative of Y
  X[i+1] = X[i] + (dxdt)*delta_t # Forward Euler method for X
  Y[i+1] = Y[i] + (dydt)*delta_t # Forward Euler method for Y
````

# Step 3: Plot our results

We can use the same plotting skills we know from before to create plots of X and Y with respect to t. We can also plot Y vs. X, which will give us something called a Phase Plane, which I will discuss later.

````python
plt.plot(t,X,'b-')
plt.plot(t,Y,'r-')
plt.ylabel('Value')
plt.xlabel('Time')
plt.legend(['x(t)','y(t)'])
````

````python
plt.plot(X,Y,'g-')
plt.ylabel('Y')
plt.xlabel('X')
plt.title('Phase Plane')
````

# Repeating above, but using a different $\Delta t$

We will see what happnes if we change our $\Delta t$ value to something larger and something smaller

````python
# Change our delta_t to 0.1 and compare to our original
delta_t = 0.1
t = np.arange(0, total_time, delta_t)
num_step = np.size(t)
X = np.zeros(np.size(t))
Y = np.zeros(np.size(t))
X[0] = 5
Y[0] = 5

for i in range(0,num_step-1) :
  dxdt = X[i]-5*Y[i] # derivative of X
  dydt = Y[i]+5*X[i] # derivative of Y
  X[i+1] = X[i] + (dxdt)*delta_t # Forward Euler method for X
  Y[i+1] = Y[i] + (dydt)*delta_t # Forward Euler method for Y

plt.plot(t,X,'b-')
plt.plot(t,Y,'r-')
plt.ylabel('Value')
plt.xlabel('Time')
plt.legend(['x(t)','y(t)'])
````

````python
plt.plot(X,Y,'g-')
plt.ylabel('Y')
plt.xlabel('X')
plt.title('Phase Plane')
````

````python
# Change our delta_t to 0.0001 and compare to our original
delta_t = 0.0001
t = np.arange(0, total_time, delta_t)
num_step = np.size(t)
X = np.zeros(np.size(t))
Y = np.zeros(np.size(t))
X[0] = 5
Y[0] = 5

for i in range(0,num_step-1) :
  dxdt = X[i]-5*Y[i] # derivative of X
  dydt = Y[i]+5*X[i] # derivative of Y
  X[i+1] = X[i] + (dxdt)*delta_t # Forward Euler method for X
  Y[i+1] = Y[i] + (dydt)*delta_t # Forward Euler method for Y

plt.plot(t,X,'b-')
plt.plot(t,Y,'r-')
plt.ylabel('Value')
plt.xlabel('Time')
plt.legend(['x(t)','y(t)'])
````

````python
plt.plot(X,Y,'g-')
plt.ylabel('Y')
plt.xlabel('X')
plt.title('Phase Plane')
````

We can see that when we make $\Delta t$ larger, our plots look super rigid, but when we make our $\Delta t$, our plots look the same as before.

Therefore, it is important to choose a value for $Δ t$ that isn't too larger, but not ridiculously small

# The Hill Equation

The Hill Equation is a mathematical model that describes the binding of a ligand to an enzyme as a function of the ligand concentration. This is very useful in understanding biochemical reactions, such as cooperativity of an enzyme or activity of a transcription factor.

The Hill Equation is given as:

$$
\nu = \frac{V_{\rm max}[S]^h}{K_{1/2}^h + [S]^h}
$$

where $\nu$ is the reaction velocity, $V_{\rm max}$ is the maximum reaction rate, $[S]$ is the ligand concentration , $K_{1/2}$ is the substrate concentration giving half-max velocity, and $h$ is the hill coefficient. When $h=1$, the Hill equation becomes identical
to the Michaelis-Menten equation.
