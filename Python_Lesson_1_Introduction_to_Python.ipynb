# Step 2: Assigning Varables
To do any of the fun programming, we need to store all our information in a way that is easily accessible. We can do this by using variables, which are names that are associated with a given bit of information

To assign a variable, we use the '=' operator

The value of our variable can either be a number, an array, or arithmetic expressions. We can also use variables to define other variables

The ```numpy``` library also includes various functions to create variables that store arrays of numbers. Some of these are:

* ```np.array([inputarray])```: stores an input array of numbers
* ```np.linspace(xmin,xmax,num)```: stores an array of ```num``` evenly-spaced values from ```xmin``` to ```xmax```
* ```np.arange(xmin,xmax,step)```: stores on array of evenly-spaced values from ```xmin``` to ```xmax``` in ```step``` intervals
* ```np.zeros(num)```: stores an array of ```num``` values, all equal 0
* ```np.ones(num)```: stores an array of ```num``` values, all equal 1

You can name variables whatever you want, but it's best to make your variable names pratical and easy to distinguish. Typical names include a series of letters, numbers and underscores. Other general rules for naming include:


*   You CANNOT start a variable name with a number
*   You CANNOT use spaces for a varable name

# Arithmetic with Variables

We can do basic mathematical operations with variables. These operations are similar to typical math notation.

* ```a + b ``` for addition;
* ```a - b``` for subraction;
* ```a * b``` for multiplication;
* ```a / b``` for division;
* ```a ** b``` for powers ($a^b$).

Other mathematical operations are implemented in ```numpy``` functions that look a bit less like standard mathematical notation:
* ```np.sqrt(a)```: evaluates $\sqrt{a}$;
* ```np.exp(x)```: evaluates $e^{x}$;
* ```np.power(a,b)```: another way to take powers ($a^b$).

# Plotting in Python

This is where things get much more intersting as we can create really nice plots of our data and our functions that we will use over the summer using simple lines of code. To plot functions, we will consider a plot of some dependent variable on the y-axis against some independent variable on the x-axis

The ```matplotlib.python``` library includes all the necessary functions we need to generate our plots. Some important fuctions include:


*   ```plt.plot(xdata,ydata)```: generates a plot y vs. x with scaled axes
* ```plt.title('title')```: creates a title for your plot at the top of the figure
*   ```plt.xlabel('title')```and```plt.ylabel('title')```: creates a label on the x and y axis respectively
* ```plt.legend(['plot 1'],['plot 2])```: creates a legend of all plots if multiple plots are a single figure
* ```plt.xlim(lower_value,upper_value)``` and ```plt.ylim(lower_value,upper_value)```: sets bounds for each axis

Other things you can do to plots is change the line color, shape, and thickness. These can be changed as followed:

```plt.plot(xdata,ydata,'(color abbreviation)(line shape)(point shape)')```

or

```plt.plot(xdata,ydata, color = 'color', linestyle = 'linestyle', marker = 'marker', linewidth = 'number')```

```Example: plot.plot(x,y,'b-o')```

This will generate a plot with a blue, solid colored line with circle points at each point

```Example2: plot.plot(x,y, color = 'red', linestyle = 'dashed', size = 3)```

This will generate a plot with a dashed red  line with a line thickness of size 3


````python
# Set up variables

x_values = np.linspace(0,10,21)
y_linear = x_values
y_squared = np.power(x_values,2)
y_sqrt = np.sqrt(x_values)

# Plot our functions
plt.plot(x_values,y_linear,'b-o')
plt.plot(x_values,y_squared, color = 'red', linestyle = 'dashed', linewidth = 3)
plt.plot(x_values,y_sqrt,'g:*')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Y vs. X')
plt.legend(['Linear','Quadratic','Square Root'])
plt.xlim(0,10)
plt.ylim(0,10)
````
