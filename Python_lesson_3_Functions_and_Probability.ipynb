# 1. Functions

In coding, a function is a block of code that can be "called" anywhere else in the code. These are very useful in reducing the amount of code needed to run simulations, especially in complicated systems.

To create a function, you will first start by defining it by using ```def```.
Then, put the name of the function to whatever you want. In most of our cases, we need a set of inputs, which will go inside parentheses right next to the function name you defined.

So all together, it should look something like this:

``` def function(input1, input2,...)```

Once your function is defined, you can write out the code for what you want your function to do

Finally, we will need for the function to give us values after it has been called, which can be done using ```return```

## Example with functions

We will do an example with the following math function.

$$ y = cos(x)*(3x^2 - 5x + 2)  $$

``` EX 1: def f_x (x) ```

```y = np.cos(x)*(3*x**2 - 5*x + 2)```

``` return y```

Now we'll see how it works in actual code

````python
import numpy as np
import matplotlib.pyplot as plt

def f_x (x):

  y = np.cos(x)*(3*x**2 - 5*x + 2)

  return y

x = np.arange(0, 100, 0.1)
y = f_x(x)

plt.plot(x, y)
````

# 3. Probability in Python

Many of the probability features are found in numpy. To generate a random number, we use ```np.random.rand()```. You can put the amount of random numbers inside the parentheses.

Note: This command will only give you a random number between 0 and 1 (Uniform Distribution), which can also be done using ```np.random.uniform()```

You can mess around with other probability distributions (Normal, Binomial, Poisson, Exponential, etc.) by replacing "rand" for any of these distributions

The best way to visualize these random numbers is through for loops and histograms. To create a histogram, we use ```plt.hist(x, binsize = #)```

Let's see this in action!

````python
for i in range(1,10):
  print(np.random.rand())

random_aarray = np.random.rand(1000)
plt.hist(random_aarray,bins=40) # change the bin size to see how the plot changes
plt.xlabel('Value')
plt.ylabel('Frequency')

## Run the code a few times to see how different each result can be ##
````

We can combine for loops, if statements, and probability all together to create fun simulations! Here's an example:

Initialize a counter variable to 0.
Then, in a loop of 1000 steps:
* Generate two random values between 0 and 1 (uniformly distributed), $a$ and $b$.
* Compute $c^2 = a^2 + b^2$, then do the following:
* If $c^2 <= 1$, plot $(a,b)$ as a blue dot, and add 1 to the counter variable.
* If $c^2 > 1$, plot $(a,b)$ as a red dot, and do *not* add anything to the counter variable.

Finally, divide your counter value by 1000, then multiply the answer by 4, and display the answer.

This may seem very strange, but we will notice a pattern.

````python
# initialize variables and placement for random points
counter = 0
x_1 = []
y_1 = []
x_2 = []
y_2 = []

for i in range (0,1000):
  a = np.random.uniform(0,1)
  b = np.random.uniform(0,1)
  c2 = a**2 + b**2

  if c2 <= 1:
    x_1.append(a)
    y_1.append(b)
    counter += 1

  else:
    x_2.append(a)
    y_2.append(b)


A = (counter/1000)*4
print(A)
plt.scatter(x_1,y_1, color='blue')
plt.scatter(x_2,y_2, color='red')
````
