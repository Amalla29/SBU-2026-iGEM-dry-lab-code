```python
# --------------- ODEs for our key proteins ---------------------

# import libraries
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
k_syn_nuc1 = 365.691 # copies/min
gamma_nuc1 = 0.03 # units: 1/min
k_syn_dspB = 296.33
k_syn_dspB_y = 260
gamma_dspB = 0.03
k_syn_sspa = 313.66 # can change to 261.44 if translation rate was 10/min
gamma_sspa = 0.03

time = 180
delta_t = 0.01
time_vec = np.arange(0, time, delta_t)
n_step = np.size(time_vec)

# Initialize variables
nuc1 = np.zeros(n_step)
dspB = np.zeros(n_step)
dspB_y = np.zeros(n_step)
sspA = np.zeros(n_step)

# Run for loop
for i in range(n_step - 1):
  dnuc1_dt = k_syn_nuc1 - gamma_nuc1*nuc1[i]
  ddspB_dt = k_syn_dspB - gamma_dspB*dspB[i]
  ddspB_y_dt = k_syn_dspB_y - gamma_dspB*dspB_y[i]
  dsspa_dt = k_syn_sspa - gamma_sspa*sspA[i]

  nuc1[i+1] = nuc1[i] + dnuc1_dt*delta_t
  dspB[i+1] = dspB[i] + ddspB_dt*delta_t
  dspB_y[i+1] = dspB_y[i] + ddspB_y_dt*delta_t
  sspA[i+1] = sspA[i] + dsspa_dt*delta_t

# Plot results
figure = plt.figure(figsize=(8, 5))
print(f"Final Nuc1 count: {np.round(nuc1[n_step-1],2)} copies")
print(f"Final dspB count: {np.round(dspB[n_step-1],2)} copies")
print(f"Final dspB_y count: {np.round(dspB_y[n_step-1],2)} copies")
print(f"Final sspa count: {np.round(sspA[n_step-1],2)} copies")
plt.plot(time_vec, nuc1, label = 'Nuc1 Protein ', color = 'blue')
plt.plot(time_vec, dspB, label = 'dspB protein', color = 'green')
plt.plot(time_vec, dspB_y, label = 'dspB_y protein', color = 'orange')
plt.plot(time_vec, sspA, label = 'sspA Protein ', color = 'red')
plt.xlabel('Time (min)')
plt.ylabel('Copies')
plt.title('Protein Copies Vs. Time')
plt.grid(alpha = 0.3)
plt.legend()
plt.show()
```

```python
# Code for Nuc1 ODEs

# import libraries
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
k_syn_nuc1 = 365.691 # copies/min
gamma_nuc1 = 0.03 # units: 1/min

time = 180
delta_t = 0.01
time_vec = np.arange(0, time, delta_t)
n_step = np.size(time_vec)

# Initialize variables
nuc1 = np.zeros(n_step)

# Run for loop
for i in range(n_step - 1):
  dnuc1_dt = k_syn_nuc1 - gamma_nuc1*nuc1[i]

  nuc1[i+1] = nuc1[i] + dnuc1_dt*delta_t

target_concentration = 12189.7 *0.9
time_at_target = 1
for i in range(n_step):
    if nuc1[i] >= target_concentration:
        time_at_target = time_vec[i]
        break

if time_at_target != 1:
    print(f"Nuc1 protein concentration reaches {target_concentration} nM at time: {np.round(time_at_target, 2)} min")
# Plot results
print(f"Final Nuc1 concentration: {np.round(nuc1[n_step-1],2)} nM")
plt.plot(time_vec, nuc1, label = 'Nuc1 Protein ', color = 'blue')
plt.xlabel('Time (min)')
plt.ylabel('Copies')
plt.title('Nuc1 ODEs')
plt.show()
```
